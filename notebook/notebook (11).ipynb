## Preparação de Dados

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv('/datasets/car_data.csv')

print('Dimensões do dataset:')
print(data.shape)

print('\nPrimeiras linhas:')
display(data.head())

print('\nInformações do dataset:')
data.info()

print('\nEstatísticas descritivas:')
display(data.describe())

print('\nValores ausentes por coluna:')
print(data.isna().sum())

print('\nNúmero de registros duplicados:')
print(data.duplicated().sum())
```

O conjunto de dados possui 354.369 registros e 16 colunas, incluindo a variável alvo Price, que representa o preço dos veículos em euros. Durante a análise, identifiquei os seguintes problemas:

Valores ausentes: as colunas VehicleType, Gearbox, Model, FuelType e NotRepaired apresentam valores ausentes. A coluna NotRepaired possui a maior quantidade, com 71.154 registros.

Registros duplicados: foram encontrados 262 registros duplicados.

Valores inconsistentes: a coluna RegistrationYear apresenta anos entre 1000 e 9999, enquanto Power possui valores entre 0 e 20.000, indicando possíveis erros nos dados.

Preços zerados: a coluna Price possui valor mínimo igual a zero, o que exige tratamento, pois o objetivo é prever o valor de mercado dos veículos.

Colunas pouco relevantes: NumberOfPictures apresenta somente valores iguais a zero. As colunas de datas e outras informações administrativas também precisam ser avaliadas quanto à sua utilidade para a previsão.

Antes do treinamento, será necessário tratar os valores ausentes, remover registros duplicados e corrigir ou excluir valores inconsistentes.

<div class="alert alert-block alert-success">
<b>Comentário do revisor — visão geral dos dados:</b> <a class="tocSkip"></a>

O diagnóstico está completo e com números: 262 duplicados, ausentes em cinco colunas (71.154 em `NotRepaired`), anos de 1000 a 9999, potência até 20.000 e `NumberOfPictures` sempre zero. Identificar esses problemas antes de tratar é o que justifica cada decisão da célula seguinte.
</div>

```python
# Analise de valores inconsistentes

print('Preços iguais a zero:')
print((data['Price'] == 0).sum())

print('\nAnos de registro anteriores a 1900:')
print((data['RegistrationYear'] < 1900).sum())

print('\nAnos de registro posteriores a 2016:')
print((data['RegistrationYear'] > 2016).sum())

print('\nPotência igual a zero:')
print((data['Power'] == 0).sum())

print('\nPotência superior a 500:')
print((data['Power'] > 500).sum())

print('\nDistribuição dos preços:')
display(data['Price'].describe())

print('\nDistribuição dos anos de registro:')
display(data['RegistrationYear'].value_counts().sort_index())
```

A análise identificou 10.772 veículos com preço igual a zero, 66 registros com ano anterior a 1900, 14.530 registros com ano posterior a 2016 e valores inconsistentes na coluna Power. Para preparar os dados, vou remover os registros com preços zerados e anos de registro fora do intervalo de 1900 a 2016.

Na coluna Power, os valores iguais a zero ou superiores a 500 serão considerados inválidos e substituídos pela mediana dos valores válidos. Os valores ausentes nas características categóricas serão preenchidos com unknown, evitando a exclusão de uma quantidade significativa de registros. Também serão removidos os registros duplicados e as colunas administrativas que não contribuem diretamente para a estimativa do preço dos veículos.

```python
# Remover registros duplicados
data = data.drop_duplicates()

# Remover preços zerados
data = data[data['Price'] > 0].copy()

# Remover anos inconsistentes
data = data[
    (data['RegistrationYear'] >= 1900) &
    (data['RegistrationYear'] <= 2016)
].copy()

# Tratar valores inconsistentes de potência
data.loc[
    (data['Power'] == 0) | (data['Power'] > 500),
    'Power'
] = np.nan

data['Power'] = data['Power'].fillna(data['Power'].median())

# Tratar valores ausentes categóricos
categorical_columns = [
    'VehicleType',
    'Gearbox',
    'Model',
    'FuelType',
    'NotRepaired'
]

data[categorical_columns] = data[categorical_columns].fillna('unknown')

# Remover colunas administrativas
columns_to_drop = [
    'DateCrawled',
    'DateCreated',
    'LastSeen',
    'NumberOfPictures',
    'PostalCode'
]

data = data.drop(columns_to_drop, axis=1)

# Verificar resultado
print('Dimensões após o tratamento:')
print(data.shape)

print('\nValores ausentes restantes:')
print(data.isna().sum())

print('\nRegistros duplicados restantes:')
print(data.duplicated().sum())

display(data.head())
```

Após o tratamento, o conjunto de dados passou de 354.369 para 329.506 registros e 11 colunas. Foram removidos registros com preços zerados e anos de registro inconsistentes. Os valores inválidos de potência foram substituídos pela mediana, enquanto os valores ausentes nas características categóricas foram preenchidos com unknown. Também removi as colunas administrativas consideradas desnecessárias para o treinamento.

Após essas transformações, não existem valores ausentes no conjunto de dados. Porém identifiquei 25.269 registros com características idênticas. Como diferentes anúncios podem representar veículos com as mesmas características e preços, esses registros serão mantidos.

<div class="alert alert-block alert-success">
<b>Comentário do revisor — tratamento:</b> <a class="tocSkip"></a>

As decisões estão justificadas e conferidas na saída: ficaram 329.506 linhas, sem ausentes. Preencher as categorias com `unknown` preserva dezenas de milhares de linhas que a exclusão descartaria (só `NotRepaired` tem 71.154 ausentes). Manter os 25.269 registros iguais depois de remover as colunas administrativas também está bem argumentado: são anúncios diferentes de carros com as mesmas características.

Como sugestão, há dois pontos a examinar. Preços muito baixos, como 1 ou 10 euros, passam pelo filtro `Price > 0` e dificilmente são valor de mercado. `RegistrationMonth` igual a 0 é um mês inexistente, e vale dizer como foi tratado.
</div>

```python
# Aplicando One-hot encoding

from sklearn.model_selection import train_test_split

features = data.drop('Price', axis=1)
target = data['Price']

features_train, features_valid_test, target_train, target_valid_test = train_test_split(
    features,
    target,
    test_size=0.4,
    random_state=12345
)

features_valid, features_test, target_valid, target_test = train_test_split(
    features_valid_test,
    target_valid_test,
    test_size=0.5,
    random_state=12345
)

print('Treinamento:', features_train.shape)
print('Validação:', features_valid.shape)
print('Teste:', features_test.shape)
```

```python
# Aplicar One-Hot Encoding
features_train_ohe = pd.get_dummies(features_train, drop_first=True)
features_valid_ohe = pd.get_dummies(features_valid, drop_first=True)
features_test_ohe = pd.get_dummies(features_test, drop_first=True)

# Garantir que os conjuntos tenham as mesmas colunas
features_valid_ohe = features_valid_ohe.reindex(
    columns=features_train_ohe.columns,
    fill_value=0
)

features_test_ohe = features_test_ohe.reindex(
    columns=features_train_ohe.columns,
    fill_value=0
)

print('Treinamento:', features_train_ohe.shape)
print('Validação:', features_valid_ohe.shape)
print('Teste:', features_test_ohe.shape)

print('\nValores ausentes:')
print(features_train_ohe.isna().sum().sum())
print(features_valid_ohe.isna().sum().sum())
print(features_test_ohe.isna().sum().sum())
```

```python
# Padronização das caracterísitcas numéricas

from sklearn.preprocessing import StandardScaler

numeric = [
    'RegistrationYear',
    'Power',
    'Mileage',
    'RegistrationMonth'
]

features_train_ohe = features_train_ohe.copy()
features_valid_ohe = features_valid_ohe.copy()
features_test_ohe = features_test_ohe.copy()

scaler = StandardScaler()
scaler.fit(features_train_ohe[numeric])

features_train_ohe[numeric] = scaler.transform(features_train_ohe[numeric])
features_valid_ohe[numeric] = scaler.transform(features_valid_ohe[numeric])
features_test_ohe[numeric] = scaler.transform(features_test_ohe[numeric])

print('Treinamento:', features_train_ohe.shape)
print('Validação:', features_valid_ohe.shape)
print('Teste:', features_test_ohe.shape)

display(features_train_ohe[numeric].head())
```

O conjunto de dados original possuía 354.369 registros e 16 colunas. Após o tratamento dos valores ausentes, remoção de registros inconsistentes e exclusão de colunas desnecessárias, restaram 329.506 registros e 11 colunas. Os dados foram divididos em três conjuntos: treinamento (60%), validação (20%) e teste (20%).

Para os modelos que exigem características numéricas, apliquei One-Hot Encoding (OHE), resultando em 312 características. Em seguida, as características numéricas foram padronizadas utilizando StandardScaler, ajustado apenas com os dados de treinamento.

Os conjuntos estão preparados para o treinamento dos modelos de aprendizado de máquina.

<div class="alert alert-block alert-success">
<b>Comentário do revisor — divisão e codificação:</b> <a class="tocSkip"></a>

A divisão 60/20/20 está certa, e o OHE foi ajustado em cada parte com o `reindex` pelas colunas do treino, o que impede colunas diferentes entre os conjuntos. O `StandardScaler` foi ajustado só no treino. Usar o tipo `category` nativo no LightGBM, em vez das 312 colunas do OHE, é a escolha adequada para essa biblioteca.
</div>

## Treinamento do modelo

```python
# Regressão Linear

import time

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Criar modelo
model_lr = LinearRegression()

# Medir tempo de treinamento
start_time = time.time()

model_lr.fit(features_train_ohe, target_train)

training_time_lr = time.time() - start_time

# Medir tempo de predição
start_time = time.time()

predicted_valid_lr = model_lr.predict(features_valid_ohe)

prediction_time_lr = time.time() - start_time

# Calcular REQM
rmse_lr = mean_squared_error(target_valid, predicted_valid_lr) ** 0.5

print('REQM:', rmse_lr)
print('Tempo de treinamento:', training_time_lr, 'segundos')
print('Tempo de predição:', prediction_time_lr, 'segundos')
```

O modelo de Regressão Linear apresentou REQM de 2.654,46 euros no conjunto de validação. O tempo de treinamento foi de aproximadamente 6,15 segundos, enquanto o tempo de predição foi de 0,12 segundos. Esses resultados serão utilizados como referência para comparar a qualidade e a velocidade dos modelos Random Forest e LightGBM.
O objetivo dos próximos treinamentos será reduzir o REQM, considerando também o tempo necessário para treinar os modelos e gerar previsões.

<div class="alert alert-block alert-warning">
<b>Comentário do revisor — Regressão Linear:</b> <a class="tocSkip"></a>

A regressão linear como referência está correta: REQM de 2.654,46, que os modelos de árvore precisam superar.

Os tempos do texto não batem com a saída da célula. O texto diz 6,15 s de treino e 0,12 s de predição, e a saída mostra 7,68 s e 0,10 s. Provavelmente o notebook foi executado de novo e os tempos mudaram, o que é normal. Como sugestão, monte o texto com f-string a partir das variáveis, para que ele acompanhe a última execução.
</div>

```python
# Random Forest

from sklearn.ensemble import RandomForestRegressor

best_model_rf = None
best_rmse_rf = float('inf')
best_depth_rf = 0
best_est_rf = 0
best_training_time_rf = 0
best_prediction_time_rf = 0

for est in [20, 50]:
    for depth in [5, 10]:
        model = RandomForestRegressor(
            n_estimators=est,
            max_depth=depth,
            random_state=12345,
            n_jobs=-1
        )

        start_time = time.time()
        model.fit(features_train_ohe, target_train)
        training_time = time.time() - start_time

        start_time = time.time()
        predicted_valid = model.predict(features_valid_ohe)
        prediction_time = time.time() - start_time

        rmse = mean_squared_error(target_valid, predicted_valid) ** 0.5

        print('Árvores:', est, '| Profundidade:', depth, '| REQM:', rmse)

        if rmse < best_rmse_rf:
            best_model_rf = model
            best_rmse_rf = rmse
            best_depth_rf = depth
            best_est_rf = est
            best_training_time_rf = training_time
            best_prediction_time_rf = prediction_time

print('\nMelhor REQM:', best_rmse_rf)
print('Melhor profundidade:', best_depth_rf)
print('Melhor número de árvores:', best_est_rf)
print('Tempo de treinamento:', best_training_time_rf, 'segundos')
print('Tempo de predição:', best_prediction_time_rf, 'segundos')
```

Foram testadas quatro combinações de hiperparâmetros, variando o número de árvores e a profundidade máxima.
O melhor resultado foi obtido com 50 árvores e profundidade máxima de 10, alcançando REQM de 1.935,04 euros no conjunto de validação. O tempo de treinamento foi de aproximadamente 37,49 segundos, enquanto o tempo de predição foi de 0,18 segundos. Em comparação com a Regressão Linear, que apresentou REQM de 2.654,46 euros, a Random Forest reduziu o erro em aproximadamente 27,1%.
Apesar da melhoria na qualidade das previsões, o modelo apresentou maior tempo de treinamento. Esses resultados serão comparados com os do LightGBM para avaliar o equilíbrio entre qualidade e desempenho.

<div class="alert alert-block alert-warning">
<b>Comentário do revisor — Random Forest:</b> <a class="tocSkip"></a>

Quatro combinações de número de árvores e profundidade, com redução de 27,1% no REQM em relação à regressão linear (1.935,04). O mesmo ponto dos tempos: o texto diz 37,49 s de treino, e a saída mostra 39,99 s.

Como sugestão, a melhor combinação ficou no limite da grade (profundidade 10). Quando o melhor valor está na borda, vale testar também 15 ou 20, para saber se o erro continua caindo.
</div>

```python
# LightGBM

from lightgbm import LGBMRegressor

# Preparar características categóricas
features_train_lgbm = features_train.copy()
features_valid_lgbm = features_valid.copy()
features_test_lgbm = features_test.copy()

categorical_columns = [
    'VehicleType',
    'Gearbox',
    'Model',
    'FuelType',
    'Brand',
    'NotRepaired'
]

for column in categorical_columns:
    features_train_lgbm[column] = features_train_lgbm[column].astype('category')
    features_valid_lgbm[column] = features_valid_lgbm[column].astype('category')
    features_test_lgbm[column] = features_test_lgbm[column].astype('category')

# Testar hiperparâmetros
best_model_lgbm = None
best_rmse_lgbm = float('inf')
best_est_lgbm = 0
best_rate_lgbm = 0
best_training_time_lgbm = 0
best_prediction_time_lgbm = 0

for est in [100, 200]:
    for rate in [0.05, 0.1]:
        model = LGBMRegressor(
            n_estimators=est,
            learning_rate=rate,
            random_state=12345,
            verbosity=-1
        )

        start_time = time.time()
        model.fit(features_train_lgbm, target_train)
        training_time = time.time() - start_time

        start_time = time.time()
        predicted_valid = model.predict(features_valid_lgbm)
        prediction_time = time.time() - start_time

        rmse = mean_squared_error(target_valid, predicted_valid) ** 0.5

        print('Árvores:', est, '| Taxa:', rate, '| REQM:', rmse)

        if rmse < best_rmse_lgbm:
            best_model_lgbm = model
            best_rmse_lgbm = rmse
            best_est_lgbm = est
            best_rate_lgbm = rate
            best_training_time_lgbm = training_time
            best_prediction_time_lgbm = prediction_time

print('\nMelhor REQM:', best_rmse_lgbm)
print('Melhor número de árvores:', best_est_lgbm)
print('Melhor taxa de aprendizado:', best_rate_lgbm)
print('Tempo de treinamento:', best_training_time_lgbm, 'segundos')
print('Tempo de predição:', best_prediction_time_lgbm, 'segundos')
```

Foram testadas quatro combinações de hiperparâmetros, variando o número de árvores e a taxa de aprendizado. O melhor resultado foi obtido com 200 árvores e taxa de aprendizado de 0,1, alcançando REQM de 1.612,35 euros no conjunto de validação. O tempo de treinamento foi de aproximadamente 3,40 segundos, enquanto o tempo de predição foi de 0,50 segundos. O LightGBM apresentou uma redução de aproximadamente 16,7% no REQM em relação à Random Forest e de 39,3% em relação à Regressão Linear. Além disso, apresentou o menor tempo de treinamento entre os modelos avaliados, embora seu tempo de predição tenha sido maior.
Portanto, o LightGBM apresentou o melhor equilíbrio entre qualidade das previsões e tempo de treinamento nesta comparação.

<b>Comentário do revisor — LightGBM:</b> <a class="tocSkip"></a>

O melhor resultado foi 1.612,35 com 200 árvores e taxa de 0,1, e as porcentagens de redução (16,7% e 39,3%) conferem. Aqui também o tempo do texto (3,40 s) difere da saída (3,92 s), sem mudar a conclusão: continua sendo o treino mais rápido.
